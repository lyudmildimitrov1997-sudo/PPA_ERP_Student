Colab ви дава възможност да се възползвате от пълната мощ на популярни библиотеки на Python, за да анализирате и визуализирате данни. Клетката с код по-долу използва <strong>numpy</strong>, за да генерира произволни данни, и <strong>matplotlib</strong>, за да ги визуализира. За да редактирате кода, трябва само да кликнете върху клетката.

В бележниците в Colab можете да импортирате собствени данни от профила си в Google Диск, включително от електронни таблици, както и от GitHub и много други източници. За да научите повече за импортирането на данни и за това как можете да използвате Colab за обработване и анализ на информация, вижте връзките под <a href="#working-with-data">Работа с данни</a> по-долу.

```
import numpy as np
import IPython.display as display
from matplotlib import pyplot as plt
import io
import base64

ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))]

fig = plt.figure(figsize=(4, 3), facecolor='w')
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6)
plt.title("Sample Visualization", fontsize=10)

data = io.BytesIO()
plt.savefig(data)
image = F"data:image/png;base64,{base64.b64encode(data.getvalue()).decode()}"
alt = "Sample Visualization"
display.display(display.Markdown(F"""![{alt}]({image})"""))
plt.close(fig)
```

```
import statsmodels.api as sm
```

```
subset_q1152_drugstores = data[(data['Product'] == 'q1152') & (data['MarketDescription_global'] == 'DRUGSTORES')]
```

```
subset_q1152_drugstores.shape
```

```
mod_q1152_drugstores = sm.OLS(subset_q1152_drugstores['VolumePC_sum'],
                              subset_q1152_drugstores['pricePerPieceNonPromo'])
```

```
mod_q1152_drugstores.fit()
```

```
mod_q1152_drugstores.fit().summary()
```

```
x = subset_q1152_drugstores['pricePerPieceNonPromo']
```

```
x = sm.add_constant(x)
```

```
x
```

```
mod_q1152_drugstores_with_constant = sm.OLS(subset_q1152_drugstores['VolumePC_sum'], x)
```

```
mod_q1152_drugstores_with_constant.fit().summary()
```

```
x = subset_q1152_drugstores[['pricePerPieceNonPromo', 'isPromoPeriod', 'TV', 'BB', 'DI']]
```

```
x = sm.add_constant(x)
```

```
x
```

```
x = x.astype(float)
```

```
mod_q1152_drugstores_with_constant = sm.OLS(subset_q1152_drugstores['VolumePC_sum'], x)
```

```
mod_q1152_drugstores_with_constant.fit().summary()
```

```
results = mod_q1152_drugstores_with_constant.fit()
predictions = results.predict(x)
```

```
pyplot.plot(subset_q1152_drugstores['Date'], subset_q1152_drugstores['VolumePC_sum'])
pyplot.plot(subset_q1152_drugstores['Date'], predictions, c='orange')
```
